```r
# ====
# APM1205 Formative Assessment 2
# Linear Models
# ====

# Load data
advertising <- read.csv("https://www.statlearning.com/s/Advertising.csv")

# Fix column name: change lowercase 'sales' to uppercase 'Sales' to match the model
colnames(advertising)[colnames(advertising) == "sales"] <- "Sales"

# Inspect data
head(advertising)
dim(advertising)
summary(advertising)

# PART 2: Scatterplot
plot(advertising$TV,
     advertising$Sales,
     main = "Sales vs. TV Advertising",
     xlab = "TV Advertising Budget ($000)",
     ylab = "Sales (000 units)",
     pch = 19)

# PART 3: Simple linear regression
model <- lm(Sales ~ TV, data = advertising)

# Regression results
summary(model)

# Regression coefficients
coef(model)

# PART 4: Fitted values & Residuals
advertising$Predicted_Sales <- fitted(model)
advertising$Residual <- residuals(model)

# Show the first 10 observations
head(advertising[, c("TV", "Sales", "Predicted_Sales", "Residual")], 10)

# PART 5: Residual plot
plot(model,
     which = 1,
     main = "Residuals vs. Fitted Values")

# PART 6: Residual standard error
summary(model)$sigma

# PART 7: Sums of squares
y <- advertising$Sales
yhat <- fitted(model)
TSS <- sum((y - mean(y))^2)
RSS <- sum((y - yhat)^2)
RegSS <- TSS - RSS

TSS
RSS
RegSS
RegSS + RSS

# PART 8: Correlation and r-squared
r <- cor(advertising$TV, advertising$Sales)
r

r_squared <- r^2
r_squared

summary(model)$r.squared
```

**PART 1** - Understanding the Dataset

**Predictor variable X**: TV advertising budget, measured in thousands of dollars.

**Response variable Y**: Product sales, measured in thousands of units.

**Research question**: Can product sales be predicted from TV advertising expenditure?.

**PART 2 - Exploratory Data Analysis**

B. Interpret the scatterplot: The relationship between TV advertising and Sales is positive and moderately strong. The form appears to follow an approximately linear upward trend, suggesting that TV advertising expenditure is indeed a potentially useful variable for predicting Sales.

**PART 3** - Simple Linear Regression Model

B. Report the coefficients: Intercept $A$ is 7.0326 and slope $B$ is 0.0475.

C. Write the regression equation: $\text{Sales} = 7.0326 + 0.0475(\text{TV})$.

D. Interpret the slope: For every additional $1,000 spent on TV advertising, predicted sales increase by approximately 0.0475 thousand units (or 47.5 units), on average.

**PART 4** - Least-Squares and Residual Analysis

C. Interpret one residual: Looking at the first observation (TV = 230.1, Actual Sales = 22.1), the residual is 4.129. This means the actual sales for this market were 4.129 thousand units higher than what the regression line predicted.

**PART 5** - Residual Plot

The residuals appear to be generally centered around zero without a severe systematic curve. There may be slight unequal spread (heteroscedasticity) as the variance widens slightly at higher fitted values, but a linear model is generally an appropriate starting point.

**PART 6** - Residual Standard Error

Value: 3.259.

Interpretation: The actual observed sales deviate from the predicted regression line by an average of 3.259 thousand units.

PART 7 - Sums of Squares
TSS: 5417.15
RSS: 2102.53
RegSS: 3314.62
Verification: $\text{RegSS} + \text{RSS} = 5417.15$, which matches the TSS exactly.

**PART 8** - Correlation and $r^2$

Correlation $r$: 0.782. The positive sign indicates a positive linear association, and the value of 0.782 shows that this relationship is moderately strong.

Coefficient of determination $r^2$: 0.612. Approximately 61.2% of the variation in Sales is explained by the TV advertising budget.

**PART 9 - Overall Conclusion**

Yes, the product sales can be predicted from the TV advertising expenditure. The moderately strong positive correlation ($r = 0.782$) and the scatterplot confirm that a clear upward linear trend. The fitted regression equation shows that sales increase by roughly 47.5 units for every additional thousand dollars spent on TV ads. Furthermore, Which the model explains 61.2% of the total variation in sales ($r^2$). While there is an average prediction error of 3.259 thousand units ($S_e$), the TV advertising remains a highly useful and statistically significant predictor of overall product sales.
